```python
import numpy as np
import os
import neurokit2 as nk
from tqdm import tqdm
import matplotlib.pyplot as plt
import pickle
import random

# ==========================================
# 0. 시드 고정
# ==========================================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)

# ==========================================
# 1. 설정
# ==========================================
TOTAL_N  = 10000
TRAIN_N  = 7000   # 70%
VAL_N    = 2000    # 20%
TEST_N   = 1000    # 10%
SEQ_LEN  = 1000
SAVE_DIR = './Demo'
os.makedirs(SAVE_DIR, exist_ok=True)
# ✅ config마다 norm_range 개별 지정
FILENAME_MAP = {
    'SKZFC_ART_VitalDB': {'tr': 'VitalDBTrART', 'val': 'VitalDBValART', 'test': 'VitalDBTestART', 'sig_type': 'ART', 'norm_range': (0.1, 0.6)},
    'SKZFC_ART_Mimic':   {'tr': 'Mimic3TrART',  'val': 'Mimic3ValART',  'test': 'Mimic3TestART',  'sig_type': 'ART', 'norm_range': (0.15, 0.55)},
    'SKZFC_II_VitalDB':  {'tr': 'VitalDBTrII',  'val': 'VitalDBValII',  'test': 'VitalDBTestII',  'sig_type': 'ECG', 'norm_range': (0.2, 0.35)}, 
    'SKZFC_II_Mimic':    {'tr': 'Mimic3TrII',   'val': 'Mimic3ValII',   'test': 'Mimic3TestII',   'sig_type': 'ECG', 'norm_range': (0.28, 0.48)}, 
}
# ==========================================
# 2. 가짜 생체 신호 생성 함수
# ==========================================
def normalize_minmax(sig, feature_range=(0, 1)):
    s_min, s_max = sig.min(), sig.max()
    if s_max - s_min == 0:
        return np.full_like(sig, feature_range[0], dtype=np.float32)
    a, b = feature_range
    return (sig - s_min) / (s_max - s_min) * (b - a) + a
def generate_dummy_signals(sig_type, num_samples, seq_len, norm_range=(0, 1)):
    signals = []
    sampling_rate = 100
    for _ in tqdm(range(num_samples), desc=f"{sig_type} 생성 중"):
        hr = np.random.randint(60, 95)
        if sig_type == 'ECG':
            sig = nk.ecg_simulate(length=1000, sampling_rate=sampling_rate, heart_rate=hr, method="simple")
        else:
            sig = nk.ppg_simulate(duration=1000 / sampling_rate, sampling_rate=sampling_rate, heart_rate=hr)
        if len(sig) > seq_len:
            sig = sig[:seq_len]
        elif len(sig) < seq_len:
            sig = np.pad(sig, (0, seq_len - len(sig)), 'constant')
        sig = normalize_minmax(sig, feature_range=norm_range)
        signals.append(sig)
    return np.array(signals, dtype=np.float32)
# ==========================================
# 3. 데이터 생성 및 분할/저장 루프
# ==========================================
assert TRAIN_N + VAL_N + TEST_N == TOTAL_N, "TRAIN_N + VAL_N + TEST_N이 TOTAL_N과 일치해야 합니다."
for ConfigName, cfg in FILENAME_MAP.items():
    tr_name   = cfg['tr']
    val_name  = cfg['val']
    test_name = cfg['test']     # ← 추가
    sig_type  = cfg['sig_type']
    norm_range = cfg['norm_range']
    print(f"\n{'='*60}")
    print(f"[{ConfigName}] 데모 데이터({sig_type}) {TOTAL_N}개 생성 시작  |  정규화 범위: {norm_range}")
    print(f"{'='*60}")
    all_dummy_signals = generate_dummy_signals(sig_type, TOTAL_N, SEQ_LEN, norm_range=norm_range)
    tr_signals   = all_dummy_signals[:TRAIN_N]
    val_signals  = all_dummy_signals[TRAIN_N:TRAIN_N + VAL_N]
    test_signals = all_dummy_signals[TRAIN_N + VAL_N:]    # ← 추가
    tr_path   = os.path.join(SAVE_DIR, f'{tr_name}.npy')
    val_path  = os.path.join(SAVE_DIR, f'{val_name}.npy')
    test_path = os.path.join(SAVE_DIR, f'{test_name}.npy')   # ← 추가
    np.save(tr_path,   tr_signals)
    np.save(val_path,  val_signals)
    np.save(test_path, test_signals)   # ← 추가
    print(f"  [저장 완료]  실제 값 범위: [{tr_signals.min():.4f}, {tr_signals.max():.4f}]")
    print(f"  Train ({TRAIN_N}개) → {tr_path}   shape: {tr_signals.shape}")
    print(f"  Val   ({VAL_N}개)  → {val_path}   shape: {val_signals.shape}")
    print(f"  Test  ({TEST_N}개) → {test_path}  shape: {test_signals.shape}")  # ← 추가

```

```python
# ==========================================
# 4. Demo 소용량 데이터 생성 (Val에서 앞 50개 슬라이싱)
# ==========================================
DEMO_N = 50

for ConfigName, cfg in FILENAME_MAP.items():
    val_name  = cfg['val']
    val_path  = os.path.join(DEMO_DIR, f'{val_name}.npy')

    val_data  = np.load(val_path)
    demo_data = val_data[:DEMO_N]

    demo_name = f"Demo_{val_name.replace('Val', '')}"
    demo_path = os.path.join(DEMO_DIR, f'{demo_name}.npy')
    np.save(demo_path, demo_data)

    print(f"[{ConfigName}] Demo 저장 완료 → {demo_path}  shape: {demo_data.shape}")
```

### 최소/최대값 불러들이기

```python
import pandas as pd
import os

# 1. 경로 및 파일 설정
load_dir = './ProcessedData'
save_dir = './Demo'
file_names = ['VitalDBSigMax', 'VitalDBSigMin', 'Mimic3SigMax', 'Mimic3SigMin']


# 저장 디렉토리가 없으면 생성
os.makedirs(save_dir, exist_ok=True)

# 2. 데이터 전체 불러오기
data_dict = {}
print("[ 데이터 로드 시작 ]")
for name in file_names:
    load_path = os.path.join(load_dir, f"{name}.pkl")
    try:
        data_dict[name] = pd.read_pickle(load_path)
        print(f"✅ Loaded: {load_path}")
    except FileNotFoundError:
        print(f"❌ Error: {load_path} 파일을 찾을 수 없습니다.")

data_dict['Mimic3SigMax'].pop('PLETH', None)
data_dict['Mimic3SigMin'].pop('PLETH', None)
```

### 최소/최대값 보정하기

```python
print(data_dict['VitalDBSigMax'], data_dict['VitalDBSigMin'], data_dict['Mimic3SigMax'], data_dict['Mimic3SigMin'])


data_dict['VitalDBSigMax']['ART'] = 200.
data_dict['VitalDBSigMax']['II'] = 2.0

data_dict['VitalDBSigMin']['ART'] = 21.
data_dict['VitalDBSigMin']['II'] = -1.1

data_dict['Mimic3SigMax']['ART'] = 220.
data_dict['Mimic3SigMax']['II'] = 5.5

data_dict['Mimic3SigMin']['ART'] = 27.5
data_dict['Mimic3SigMin']['II'] = -3.0

print(data_dict['VitalDBSigMax'], data_dict['VitalDBSigMin'], data_dict['Mimic3SigMax'], data_dict['Mimic3SigMin'])
```

### Fake 최소/최대값 저장하기

```python
os.makedirs(save_dir, exist_ok=True)

for name, data in data_dict.items():
    save_path = os.path.join(save_dir, f"{name}.pkl")
    
    with open(save_path, 'wb') as f:
        pickle.dump(data, f)

print("\nAll files have been saved to /Demo/")
```
